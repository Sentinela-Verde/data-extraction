```python
import pandas as pd
```

```python
df = pd.read_csv('../data/02_silver/datacentermap_enriquecido.csv', sep=';', encoding='utf-8', low_memory=False)
```

# Definição do Range de dados do facility
- Para medir o impacto para os anos, vamos considerar uma janela movel de 4 anos. Exemplo: facility foi inaugurado em 2020. ano0 = 2020, ano1 = 2021, ano_1 = 2019 ano_2 = 2018, ano_3= 2017

```python
mask1 = df['status'] == 1 # indica se o datacenter esta ativo ou nao
mask2 = ~df['ano_operacional'].isna() #filtra casos que o ano não é nulo
mask3 = df['ano_operacional'] > 2017 #filtra casos que o ano é maior que 2017 pois fica com a amostra muito distante do estudo
mask4 = df['ano_operacional'] < 2025 #filtra casos que o ano é menor que 2025 pois preciso ter pelo menos 2026 de referencia
mask5 = df['stage'] == 2
```

```python
df_f = df[mask1 & mask2 & mask3 & mask4 & mask5].reset_index(drop=True)
df_f['ano_operacional'] = df_f['ano_operacional'].astype(int)
```

```python
df_f = df_f[['id_datacenter', 'nome_datacenter', 'endereco', 'cidade', 'estado', 'latitude', 'longitude',
             'tags', 'mw_construido', 'whitespace_construido_m', 'ano_operacional', 'tipo_construcao']]
```

tags:
Hyperscale-Ready: Refere-se a datacenters construídos para abrigar operações de empresas conhecidas como Hyperscalers (como Google, AWS, Microsoft e Meta)
- Capacidade de infraestrutura de atender a gigantes de tecnologia com demandas massivas de escala
- Alta Densidade de Energia
- Eficiência Energética: Focam em um PUE (Power Usage Effectiveness) extremamente baixo para reduzir custos e impactos ambientais.
Carrier Neutral: Significa que o datacenter opera de forma independente e não é controlado por nenhuma empresa de telecomunicações específica.

```python
df_f.to_csv('datacenter_filtrado.csv', sep=';', index=False, encoding='utf-8')
```
